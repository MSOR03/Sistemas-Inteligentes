###Juego
* Al inicio del jugo cada **agente** generara un numero aleatorio de cuatro cifras diferentes

  > 🚨 Los numeros 0000, 101 o 12345 no son numeros validos ya que, o repiten cifras, o no tienen el numero correcto de sifras.

  >Esto se hara mediante la funcion incial.

    ```
    TuAgente.start
    ```
* cada **agente** intentara adivinar el numero del contrincante una vez por turno.

  >🚨Cada **intento** es una **lista** de enteros con 4 elementos *(de 0 a 3)* donde el elemento 0 es el primer digito del numero, asi la lista [1,2,3,4] se lee 1234.

  >Esto se hara cada **turno** mediante la funcion.
    ```
    TuAgente.try
    ```
* Cada **agente** recivira el **intento** del contrincante *(despues de mandar su propio intento)* y respondera que tan acertado esta el **agente** opuesto.

  >🚨 La **Respuesta** es una lista de enteros con 2 elementos *(de 0 a 1)* donde picas y fijas son los elemntos 0 y 1 respectivamente, asi  la lista [2,1] indica que el **intento** tuvo 2 **picas** y una **fija**.

  >Esto se hara cada **turno** mediante la funcion.

    ```
    TuAgente.discover(numeroLista)
    ```
* Cada **agente** obtendra retroalimentacion de su intento cada turno *(despues de responder al contrincante)* con la respuesta del agente rival.
  >🚨La retroalimentacion vendra dada por es una lista de enteros con 2 elementos *(de 0 a 1)* donde picas y fijas son los elemntos 0 y 1 respectivamente, asi  la lista [2,1] indica que el **intento** tuvo 2 **picas** y una **fija**.

  >Esto se hara cada **turno** mediante la funcion.

    ```
    TuAgente.feedBack(retroalimentacionLista)
    ```




###Interfaz para los modelos

````python
from abc import ABC, abstractmethod

class interfazAgente(ABC):

  @abstractmethod
  def start():
    """Funcion que inicializa el agente creando el numero a adivinar"""
    pass

  @abstractmethod
  def try():
    """Funcion compute que, mediante los datos dados, intenta adivinar el numero del contrincante"""
    pass

  @abstractmethod
  def discover(numeroLista) -> list(int):
    """Funcion que dado un intento devuelve la precicion del mismo"""
    pass

  @abstractmethod
  def feedBack(retroalimentacionLista):
    """Funcion que recibe retroalimentacion del ambiente"""
    pass
````
